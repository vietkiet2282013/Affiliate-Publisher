# 🚀 HỆ THỐNG ĐĂNG VIDEO ĐA KÊNH & ĐA NỀN TẢNG

Quản lý và tự động đăng tải video cho **3 Thương hiệu (Kênh)** trên các nền tảng **TikTok, YouTube Shorts, Facebook Reels, Instagram Reels**:
1. 🌟 **Nói Thật Dùng Thật** (TikTok, YouTube, Facebook, Instagram)
2. 🌟 **Dùng Rồi Mới Nói**   (YouTube, Facebook, Instagram)
3. 🌟 **Review Thật Việc Thật** (YouTube, Facebook, Instagram)

```python
# 1. Import các thư viện và cấu hình
import os
import json
from uploaders.facebook_uploader import FacebookReelsUploader
from uploaders.youtube_uploader import YouTubeShortsUploader
from uploaders.instagram_uploader import InstagramReelsUploader

# Nạp cấu hình các kênh từ config/brands.json
with open("config/brands.json", "r", encoding="utf-8") as f:
    BRANDS = json.load(f)

print("✅ Đã nạp thành công cấu hình 3 thương hiệu:", list(BRANDS.keys()))
```

--- 
## 🏷️ KÊNH 1: DÙNG RỒI MỚI NÓI

```python
brand_dung_roi = BRANDS["dung_roi_moi_noi"]

VIDEO_PATH_1 = os.path.join("Dùng rồi mới nói", "keo_cat_ga.mp4")  # Chọn video trong folder
TITLE_1 = "Kéo Cắt Gà Đa Năng - Dùng Rồi Mới Nói"
CAPTION_1 = "Dùng rồi mới nói! Test thật kéo cắt gà siêu bén 🔥 #reels #shorts #review #fyp"

# 1. Đăng YouTube Shorts
yt_1 = YouTubeShortsUploader(token_pickle_file=brand_dung_roi["youtube"]["token_file"])
yt_1.upload_short(video_path=VIDEO_PATH_1, title=TITLE_1, description=CAPTION_1)

# 2. Đăng Facebook Reels
fb_1 = FacebookReelsUploader(page_id=brand_dung_roi["facebook"]["page_id"], access_token=brand_dung_roi["facebook"]["access_token"])
fb_1.upload_reel(video_path=VIDEO_PATH_1, caption=CAPTION_1, title=TITLE_1)
```

--- 
## 🏷️ KÊNH 2: REVIEW THẬT VIỆC THẬT

```python
brand_review_that = BRANDS["review_that_viec_that"]

VIDEO_PATH_2 = os.path.join("Review thật việc thật", "bào_rau.mp4")  # Chọn video trong folder
TITLE_2 = "Bào Rau Củ Đa Năng - Review Thật Việc Thật"
CAPTION_2 = "Nói thật, test thật, trải nghiệm thật! 🔥 Giúp bạn mua sắm thông minh #reels #shorts #review #viral"

# 1. Đăng YouTube Shorts
yt_2 = YouTubeShortsUploader(token_pickle_file=brand_review_that["youtube"]["token_file"])
yt_2.upload_short(video_path=VIDEO_PATH_2, title=TITLE_2, description=CAPTION_2)

# 2. Đăng Facebook Reels
fb_2 = FacebookReelsUploader(page_id=brand_review_that["facebook"]["page_id"], access_token=brand_review_that["facebook"]["access_token"])
fb_2.upload_reel(video_path=VIDEO_PATH_2, caption=CAPTION_2, title=TITLE_2)
```

--- 
## 🏷️ KÊNH 3: NÓI THẬT DÙNG THẬT

```python
brand_noi_that = BRANDS["noi_that_dung_that"]

VIDEO_PATH_3 = os.path.join("Nói thật dùng thật", "ke_chan.mp4")  # Chọn video trong folder
TITLE_3 = "Kệ Chân Tiện Lợi - Nói Thật Dùng Thật"
CAPTION_3 = "Nói Thật Dùng Thật! Đánh giá thực tế 🔥 #tiktok #reels #shorts #review"

# 1. Đăng YouTube Shorts
yt_3 = YouTubeShortsUploader(token_pickle_file=brand_noi_that["youtube"]["token_file"])
yt_3.upload_short(video_path=VIDEO_PATH_3, title=TITLE_3, description=CAPTION_3)

# 2. Đăng Facebook Reels (Điền page_id và token vào config/brands.json khi bạn có Page của kênh này)
if brand_noi_that["facebook"]["page_id"]:
    fb_3 = FacebookReelsUploader(page_id=brand_noi_that["facebook"]["page_id"], access_token=brand_noi_that["facebook"]["access_token"])
    fb_3.upload_reel(video_path=VIDEO_PATH_3, caption=CAPTION_3, title=TITLE_3)
```
